```python
import sqlite3

con = sqlite3.connect("movie.db")
cur = con.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS bookings (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    movie TEXT NOT NULL,
    tickets INTEGER NOT NULL,
    total REAL NOT NULL
)
""")
con.commit()

movies = {
    "1": ("Ramayan", 200),
    "2": ("Dune", 150),
    "3": ("Avengers: Doomsday", 180)
}

timings = {
    "Ramayan": "10:00 AM",
    "Dune": "2:00 PM",
    "Avengers: Doomsday": "6:00 PM"
}

while True:
    print("\n1. View Movies")
    print("2. View Show Timings")
    print("3. Book Tickets")
    print("4. View Bookings")
    print("5. Cancel Booking")
    print("6. Exit")

    choice = input("Enter choice: ").strip()

    if choice == "1":
        print("\nAvailable Movies:")

        for key, value in movies.items():
            print(key, value[0], "- Rs.", value[1])

    elif choice == "2":
        print("\nMovie Show Timings:")

        for movie, time in timings.items():
            print(movie, "-", time)

    elif choice == "3":
        name = input("Enter your name: ").strip()

        if not name:
            print("Name cannot be empty.")
            continue

        print("\nAvailable Movies:")

        for key, value in movies.items():
            print(key, value[0], "- Rs.", value[1])

        movie_id = input("Select movie: ").strip()

        if movie_id in movies:
            try:
                tickets = int(input("Number of tickets: "))

                if tickets <= 0:
                    print("Enter a valid number of tickets.")
                    continue

                movie, price = movies[movie_id]
                total = tickets * price

                cur.execute(
                    """INSERT INTO bookings
                    (name, movie, tickets, total)
                    VALUES (?, ?, ?, ?)""",
                    (name, movie, tickets, total)
                )

                con.commit()

                booking_id = cur.lastrowid

                print("\nBooking successful!")
                print("Booking ID:", booking_id)
                print("Customer:", name)
                print("Movie:", movie)
                print("Show Time:", timings[movie])
                print("Tickets:", tickets)
                print("Total Price: Rs.", total)

            except ValueError:
                print("Please enter a valid number of tickets.")

        else:
            print("Invalid movie selection.")

    elif choice == "4":
        cur.execute("SELECT * FROM bookings")
        records = cur.fetchall()

        if not records:
            print("No bookings found.")
        else:
            print("\nBooking Details:")

            for record in records:
                print("--------------------")
                print("Booking ID:", record[0])
                print("Name:", record[1])
                print("Movie:", record[2])
                print("Tickets:", record[3])
                print("Total: Rs.", record[4])

    elif choice == "5":
        try:
            booking_id = int(
                input("Enter booking ID to cancel: ")
            )

            cur.execute(
                "DELETE FROM bookings WHERE id = ?",
                (booking_id,)
            )

            if cur.rowcount > 0:
                con.commit()
                print("Booking cancelled successfully.")
            else:
                print("Booking ID not found.")

        except ValueError:
            print("Enter a valid booking ID.")

    elif choice == "6":
        print("Thank you for using the Movie Ticket Booking System!")
        break

    else:
        print("Invalid choice. Please try again.")

con.close()
```
